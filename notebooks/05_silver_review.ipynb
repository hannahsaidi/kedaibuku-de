1. Quarantine vs delete
Quarantine keeps bad rows so they can be checked and fixed. Deleting hides the problem. A data team would check the file.

2. Hash vs mask
Without a salt, hackers can guess common emails, hash them, and match the results. The secret salt stops this.

3. Data minimisation
PDPA says don’t keep more personal data than you need. A age band (25-34) is safer than birth year because it doesn’t point to one person.

4. Parquet vs CSV
Parquet remembers data types (like dates and numbers). CSV is just text, so the types get lost.
