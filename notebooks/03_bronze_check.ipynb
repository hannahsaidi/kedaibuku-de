```python
import pandas as pd

for name in["books", "customers" , "orders" , "reviews"]:
    df = pd.read_csv(f"../data/lake/bronze/{name}.csv")
    print(name, df.shape)
    display(df.head())
```

1. What does one row mean in each file?

books: one book
customers: one customer
orders: one order (one book and its quantity)
reviews: one review written by a customer for a book

2. Which columns link the files together?

customer_id links customers, orders and reviews.
book_id links books, orders and reviews.

3. Which columns hold personal data?

full_name, email, phone and mykad in customers.
customer_id in orders and reviews is indirect personal data, because it can be traced back to one person.

4. Which source would you collect once, and which every day?

Once: the book catalogue (books), because it rarely changes.
Every day: the exchange rate (FX), because it changes daily, and orders and reviews, because new ones keep arriving. customers is also collected regularly for new sign-ups.
